```python
empid=input("enter the empid : ")
empname=input("enter the name : ")
bs =float(input("enter the salary : "))
da =bs*0.08
ta =bs*0.12
ba = 3000
pf = bs*0.15
tax = bs *0.05
ns = bs+da+ta+ba
gs = ns+pf+tax
```

```python
print("*"*10 ,"salary slip", "*"*10)
print(f"empid:{empid}\nempname:{empname}\nbasic salary:{bs}")
print(f"Dearness allowance :{da}\ntotal amount:{ta}\nBonus amount{ba}")
print(f"pf:{pf}\ntax amount{ta:.2f}\nnet amount:{ns}\n gross amount{gs}")
print("*"*10, "thank you" , "*"*10)
```
