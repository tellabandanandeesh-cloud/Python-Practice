```python
write a program to print itss a equilater triangle , 
isoceles triangle , scalene triangle or not a triangle 
```

```python
a = float(input("enter a 1st side of triangle :"))
b = float(input("enter a 2nd side of triangle :"))
c = float(input("enter a 3rd side of triangle :"))
if(a+b<c or a+b<b or b+c<a):
    print("not a triangle")
elif(a==b==c):
    print("equilateral triangle")
elif(a==b or b==c or c==a ):
    print("isocelese triangle")
else:
    print("scalence triangle")
```
